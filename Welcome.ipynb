<h1> Partial derivative</h1>

A derivative tells us how quickly something is changing at a particular point.



```python
# chicken biryani cost fininding with torch tensor partical derivative
# creating two tensors one is for rice  cost and other for chicken cost 

rice_cost = torch.tensor(50, requires_grad=True,dtype=torch.float32) # rice cost
chicken_cost = torch.tensor(200, requires_grad=True,dtype=torch.float32) # chicken cost
```

```python
rice_cost
```

```python
biryani_cost = 3* rice_cost + chicken_cost
```

```python
print("Total biryani cost:", biryani_cost.item())
```

```python
# now find the partial derivative of biryani cost with respect to rice cost and chicken cost

biryani_cost.backward()
```

```python
print("Partial derivative w.r.t. rice cost:", rice_cost.grad.item())
print("Partial derivative w.r.t. chicken cost:", chicken_cost.grad.item())
```
